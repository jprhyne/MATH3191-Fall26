```python
import numpy as np
import sympy as sym
```

```python
A = np.array([[4, -2, 5, -5],
              [-9, 7, -8, 0],
              [-6, 4, 5, 3],
              [ 5, -3, 8, -5]])
print(A)
```

```python
Asym = sym.Matrix(A)
Asym.rref()
```

```python
bsym = sym.Matrix([[7],[5],[9],[7]])

A_aug_sym = Asym.col_insert(5, bsym)

A_aug_sym.rref()
```
